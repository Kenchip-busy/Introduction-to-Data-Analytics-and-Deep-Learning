1. Giới thiệu về thư viện NLTK:



```python
import nltk

nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve the error

gb = nltk.corpus.gutenberg
print("Gutenberg files: ", gb.fileids())

macbeth = nltk.corpus.gutenberg.words('shakespeare-macbeth.txt')
print("\nSố lượng từ:", len(macbeth))
print("10 từ đầu tiên:", macbeth[:10])

macbeth_sents = nltk.corpus.gutenberg.sents('shakespeare-macbeth.txt')
print("\n5 câu đầu:", macbeth_sents[:5])
```

2. Tìm 1 từ với NLTK

```python
text = nltk.Text(macbeth)

print("Các câu chứa từ 'Stage':")
text.concordance('Stage')

print("\nNgữ cảnh chung của từ 'Stage':")
text.common_contexts(['Stage'])

print("\nCác từ tương tự ngữ cảnh với 'Stage':")
text.similar('Stage')
```

3. Phân tích tần số của các từ

```python
import string

fd = nltk.FreqDist(macbeth)
print("10 từ phổ biến nhất:", fd.most_common(10))

nltk.download('stopwords')
sw = set(nltk.corpus.stopwords.words('english'))

punctuation = set(string.punctuation)
macbeth_filtered2 = [w.lower() for w in macbeth
                     if w.lower() not in sw and w.lower() not in punctuation]

fd_filtered = nltk.FreqDist(macbeth_filtered2)
print("\n10 từ phổ biến sau khi lọc:", fd_filtered.most_common(10))
```

4. Lựa chọn các từ trong văn bản

```python
long_words = [w for w in macbeth if len(w) > 12]
print("Từ dài hơn 12 ký tự:", sorted(long_words))

ious_words = [w for w in macbeth if 'ious' in w]
ious_words = set(ious_words)
print("\nTừ chứa 'ious':", sorted(ious_words))
```

5. Bigrams và collocations

```python
bgrms = nltk.FreqDist(nltk.bigrams(macbeth_filtered2))
print("15 Bigrams phổ biến nhất:")
for bg, freq in bgrms.most_common(15):
    print(f"{bg}: {freq}")

tgrms = nltk.FreqDist(nltk.trigrams(macbeth_filtered2))
print("\n10 Trigrams phổ biến nhất:")
for tg, freq in tgrms.most_common(10):
    print(f"{tg}: {freq}")
```

7. Rút trích văn bản từ trang html

```python
from urllib import request
from bs4 import BeautifulSoup
import nltk

url = "http://news.bbc.co.uk/2/hi/health/2284783.stm"
html = request.urlopen(url).read().decode('utf8')

print("120 ký tự đầu của HTML:", html[:120])

raw_html = BeautifulSoup(html, "lxml").get_text()
tokens_html = nltk.word_tokenize(raw_html)
text_html = nltk.Text(tokens_html)

print("\n10 tokens đầu tiên sau khi bóc tách:", text_html[:10])
```

8. Phân tích cảm xúc người dùng

```python
import random
import nltk

nltk.download('movie_reviews')
reviews = nltk.corpus.movie_reviews

documents = [(list(reviews.words(fileid)), category)
             for category in reviews.categories()
             for fileid in reviews.fileids(category)]
random.shuffle(documents)

all_words = nltk.FreqDist(w.lower() for w in reviews.words())
word_features = list(all_words)[:2000]

def document_features(document, word_features):
    document_words = set(document)
    features = {}
    for word in word_features:
        features[f'contains({word})'] = (word in document_words)
    return features

featuresets = [(document_features(d, word_features), c) for (d,c) in documents]

train_set, test_set = featuresets[1500:], featuresets[:500]

classifier = nltk.NaiveBayesClassifier.train(train_set)

print("Độ chính xác:", nltk.classify.accuracy(classifier, test_set))
print("\nCác đặc trưng quan trọng nhất:")
classifier.show_most_informative_features(10)
```

9. Bài tập áp dụng


```python
import nltk
import os
from nltk.corpus import stopwords, wordnet, names
from nltk.tokenize import word_tokenize
import random

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('tagsets')
nltk.download('tagsets_json') # Added to fix the LookupError
nltk.download('names')

print("--- Câu 1 ---")
corpora_dir = nltk.data.find('corpora')
print("Các corpora đã tải (Top 10):", os.listdir(corpora_dir)[:10])

print("\n--- Câu 2 & 3 ---")
print("Ngôn ngữ hỗ trợ stopwords:", stopwords.fileids())
print("Stopwords tiếng Pháp (Top 10):", stopwords.words('french')[:10])

print("\n--- Câu 4 ---")
text = "Software engineering is an exciting field for students."
tokens = word_tokenize(text)
english_sw = set(stopwords.words('english'))
filtered_text = [w for w in tokens if w.lower() not in english_sw]
print("Sau khi lọc stopwords:", filtered_text)

print("\n--- Câu 5 ---")
custom_sw = set(stopwords.words('english')) - {'not', 'no', 'against'}
print("Đã loại bỏ các từ phủ định khỏi tập stopwords.")

print("\n--- Câu 6 ---")
syn = wordnet.synsets("software")[0]
print(f"Từ 'software' - Định nghĩa: {syn.definition()}")
print(f"Ví dụ: {syn.examples()}")

print("\n--- Câu 7 ---")
synonyms, antonyms = [], []
for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())
        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())
print("Đồng nghĩa 'good':", set(synonyms))
print("Trái nghĩa 'good':", set(antonyms))

print("\n--- Câu 8 ---")
nltk.help.upenn_tagset('NN.*')

print("\n--- Câu 9 & 10 ---")
cat, dog = wordnet.synset('cat.n.01'), wordnet.synset('dog.n.01')
run, sprint = wordnet.synset('run.v.01'), wordnet.synset('sprint.v.01')
print("Độ tương đồng Noun (cat-dog):", cat.wup_similarity(dog))
print("Độ tương đồng Verb (run-sprint):", run.wup_similarity(sprint))

print("\n--- Câu 11 ---")
male_names = names.words('male.txt')
female_names = names.words('female.txt')
print(f"Số lượng nam: {len(male_names)}, nữ: {len(female_names)}")
print("10 Tên nam đầu:", male_names[:10])

print("\n--- Câu 12 ---")
labeled_names = ([(name, 'male') for name in male_names] +
                 [(name, 'female') for name in female_names])
random.shuffle(labeled_names)
print("15 mẫu ngẫu nhiên:", labeled_names[:15])

print("\n--- Câu 13 ---")
featuresets_names = [(n[-1].lower(), g) for (n, g) in labeled_names]
print("Mảng ký tự cuối (Top 10):", featuresets_names[:10])
```
